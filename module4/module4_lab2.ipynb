## Load the Data

```python
import pandas as pd

df = pd.read_csv('allbacks.csv', index_col='rownames')
print('Shape:', df.shape)
df
```

Weight is in grams, volume in cubic centimeters. `area` is 0 for every paperback in this file (it differs from the textbook table), but area is not used in this problem.

## Median Cover Type (Textbook Part A)

```python
# A median works on ordered categories: sort, then take the middle value
cover_order = pd.CategoricalDtype(['hb', 'pb'], ordered=True)
covers = df['cover'].astype(cover_order).sort_values().reset_index(drop=True)

print(df['cover'].value_counts())
print()
middle = (len(covers) - 1) // 2          # 15 books -> position 8 of 15 (index 7)
print('Middle position:', middle + 1, 'of', len(covers))
print('Median cover:', covers[middle])
```

**Analysis:** There were 7 hardback copies that come before the 8 paperback copies; therefore, the 8th book out of 15 would be one of the paperback copies. Therefore, the median cover type in this sample is paperback. If we reversed the order and put the paperbacks first, the answer would still be the same because paperbacks make up more than half of the books.

## Mean Weight (Textbook Part B)

```python
print(f"Mean weight, all books: {df['weight'].mean():.2f} g")
print()
by_cover = df.groupby('cover').agg(books=('weight', 'size'),
                                   mean_weight_g=('weight', 'mean'),
                                   mean_volume_cm3=('volume', 'mean')).round(2)
by_cover
```

```python
diff = by_cover.loc['hb', 'mean_weight_g'] - by_cover.loc['pb', 'mean_weight_g']
print(f'Hardbacks are {diff:.2f} g heavier on average ({diff / by_cover.loc["pb", "mean_weight_g"]:.0%} heavier than paperbacks)')
```

**All Books:** We can estimate that the average weight for a book in this sample is approximately 707 grams.

**By Cover Type:** The average weights for hardback books and paperback books were respectively around 796 g and 628 g. Therefore, there is an approximate difference of 168 grams (27%) between the two cover types, with hardbacks being heavier. Although both have nearly equal average volumes (hardbacks = 834 cm³; paperbacks = 856 cm³), the additional weight of the hardbacks comes from the cover type rather than the hardbacks being larger books. The library may find paperbacks beneficial because their lighter weight would make handling them and stocking shelves easier. However, using only 7 hardbacks and 8 paperbacks as a sample is too limited to base a long-term policy on.

## Variance in Book Volume (Textbook Part C)

```python
print(f"Variance of volume, all books: {df['volume'].var():,.2f} (cm³)²")
print()
df.groupby('cover')['volume'].agg(['mean', 'var', 'std']).round(2)
```

The standard deviation for book volume is approximately 340 cm³ overall. The spread is also similar for both cover types; therefore, neither hardbacks nor paperbacks have a narrower range of book sizes. The variability in book volume between the two cover types is very similar.

## Recommendation

The library should select paperback. This is the median cover type in the current sample as well as the more common one. Additionally, even though paperbacks and hardbacks are nearly the same in volume, a paperback weighs about 168 g less than a hardback. The range of book sizes is also about the same for both cover types.
