## Prepare the Features

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

titanic = pd.read_csv("titanic_train.csv")
print("Missing ages before:", titanic["Age"].isna().sum())

median_age = titanic["Age"].median()
titanic["Age"] = titanic["Age"].fillna(median_age)
titanic["Sex"] = titanic["Sex"].map({"male": 0, "female": 1})

print("Median age used:", median_age)
print("Missing ages after:", titanic["Age"].isna().sum())

X = titanic[["Pclass", "Sex", "Age", "Fare"]]
y = titanic["Survived"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Training passengers:", len(X_train))
print("Test passengers:", len(X_test))
```

I picked "passenger class," "sex", "age" & "fare" as my best options for what would probably influence if a passenger lived through the sinking of the Titanic, because they were either already numeric, or could easily be made numeric. Since "sex" was one of the variables, I had to assign it a numeric value, so I assigned the number 0 for males and 1 for females. I identified 177 people who had missing ages; therefore, I substituted the median age for those missing ages based on the ages of those passengers whose ages were available. I selected the median over the mean because it is less influenced by outliers. This enabled me to include the 177 passengers in the analysis, as opposed to eliminating them due to missing ages.

## Fit the Classifier

```python
tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(X_train, y_train)
print("Test accuracy:", round(tree.score(X_test, y_test), 3))
```

## Error Analysis

```python
results = X_test.copy()
results["Actual"] = y_test
results["Predicted"] = tree.predict(X_test)
mistakes = results[results["Actual"] != results["Predicted"]].copy()

mistakes["Passenger"] = mistakes["Sex"].map({0: "Man", 1: "Woman"})
mistakes["What happened"] = mistakes["Actual"].map({0: "Died", 1: "Survived"})

print("Total mistakes:", len(mistakes))
mistakes.groupby(["Passenger", "What happened"]).size().reset_index(name="Mistakes")
```

A total of 52 wrong predictions were generated by the model, with 33 of those being men that had been predicted to die instead of survive. As such, 63.5% of the models errors (the wrong prediction) were men who survived but were incorrectly identified as non-survivors. It appears that the model is using gender as a primary factor in determining survivorship of the passengers and therefore it is likely that the model will underestimate the potential for males to survive. Nonetheless, the fact that most of the "wrong" predictions were men does not necessarily mean that gender was the cause of the errors since there could have been numerous other contributing variables including; class, age or fare.
