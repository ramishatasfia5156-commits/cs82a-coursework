## Fit the Regression

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

ratings = pd.read_csv("ratings.tsv", sep="\t")
X = ratings[["food", "ambience", "service"]]
y = ratings["rating"]

model = LinearRegression()
model.fit(X, y)

print("Intercept:", round(model.intercept_, 3))
for name, coef in zip(X.columns, model.coef_):
    print(name + ":", round(coef, 3))
print("R-squared:", round(model.score(X, y), 3))
```

rating = -15.87 + 0.73 x food + 0.13 x ambience + 0.32 x service

## Ranked Coefficients

```python
ranked = pd.DataFrame({"feature": X.columns, "coefficient": model.coef_})
ranked = ranked.sort_values("coefficient", ascending=False).reset_index(drop=True)
ranked.index = ranked.index + 1
ranked
```

## Recommendation

Food Quality will likely be the first area for improvement for an underperforming restaurant. Food quality was found in the Multiple Regression Analysis to have the highest coefficient value (0.73) and therefore should produce the greatest increase in overall rating. Service had a coefficient value of 0.32; therefore it can produce a smaller but still positive increase in overall rating. Ambiance produced a much lower coefficient value (.13); therefore it will produce the least amount of increased in overall rating. One point increases in food quality, service, and ambiance are also related to a .73,.32 and .13 point increase respectively in overall rating when service and ambiance do not change. Therefore, based on this data, food quality produces about 2.3 times as great of an estimate impact than does service. It also produces approximately 5.5 times as great of an estimate impact than does ambiance. However, since there were only 15 total restaurants included in the regression analysis, we cannot assume that the estimates derived from this analysis apply to every restaurant.
