```python
import sqlite3
import pandas as pd

con = sqlite3.connect("lab2.db")
con.execute("DROP TABLE IF EXISTS customers")
con.execute("""
CREATE TABLE customers (
    id            INTEGER PRIMARY KEY,
    name          TEXT NOT NULL,
    city          TEXT,
    signup_year   INTEGER,
    total_spent   REAL
)
""")
print("Table created")
```

```python
con.execute("""
INSERT INTO customers (name, city, signup_year, total_spent) VALUES
    ('Nova Quill',    'Santa Monica', 2021, 412.50),
    ('Juniper Vale',  'Culver City',  2022, 138.25),
    ('Orion Pike',    'Venice',       2023, 889.00),
    ('Sable Hartwell','Inglewood',    2020, 57.80),
    ('Marlow Finch',  'Los Angeles',  2024, 1204.99),
    ('Tamsin Reeve',  'Torrance',     2022, 301.10),
    ('Ellis Moreau',  'Santa Monica', 2025, 76.40),
    ('Wren Calloway', 'Culver City',  2021, 640.00)
""")
con.commit()
print("Rows inserted:", con.execute("SELECT COUNT(*) FROM customers").fetchone()[0])
```

```python
pd.read_sql("SELECT * FROM customers", con)
```

The textbook's MySQL table declares its key as `ID int(11) NOT NULL auto_increment` with `PRIMARY KEY ('ID')`, and its insert passes an empty value for the ID so MySQL fills it in. SQLite has no `auto_increment` keyword. Instead, a column declared exactly as `INTEGER PRIMARY KEY` numbers itself automatically, so my table uses that and my insert leaves the id column out entirely. Related changes: SQLite uses flexible types like `TEXT` and `REAL` instead of `char(35)` and `int(11)`, and there is no `CREATE DATABASE` or `use` step because `sqlite3.connect()` creates the file the first time it runs.
