## Problem 6.1: Pounds to kilograms
The conversion factor is 1 lb = 0.453592 kg.

```python
# Function that converts pounds (lbs) to kilograms (kgs)
def lbs_to_kgs(lbs):
    """Return the weight in kilograms for a weight given in pounds."""
    return lbs * 0.453592
```

## Calling the function

```python
# Call 1: one pound
print("1 lb =", round(lbs_to_kgs(1), 4), "kg")

# Call 2: one hundred pounds
print("100 lbs =", round(lbs_to_kgs(100), 4), "kg")

# Call 3: my own weight in pounds (EDIT the number below before running)
my_weight_lbs = 150
print(my_weight_lbs, "lbs =", round(lbs_to_kgs(my_weight_lbs), 4), "kg")
```

```python
# Summary table of all three calls
import pandas as pd

weights_lbs = [1, 100, my_weight_lbs]
table = pd.DataFrame({
    "Pounds (lbs)": weights_lbs,
    "Kilograms (kgs)": [round(lbs_to_kgs(w), 4) for w in weights_lbs],
})
table
```
