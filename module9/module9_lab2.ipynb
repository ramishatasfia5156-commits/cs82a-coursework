## Migration note: moving my CS 82A workflow from Google Colab to Kaggle Notebooks
My course work is small notebooks (pandas, seaborn, scikit-learn, scipy) that read one CSV and produce tables and charts. I run them in Colab and keep copies in my GitHub repo. The one other platform I picked is **Kaggle Notebooks**, another free cloud notebook service.

### What transfers untouched
| Item | Why it moves as is |
|---|---|
| The `.ipynb` files | Both platforms use the standard Jupyter notebook format, so I can download from Colab or GitHub and import. |
| The Python code | pandas, seaborn, matplotlib, scikit-learn and scipy are all standard libraries, and my code only uses ordinary functions like `groupby`, `StandardScaler` and `linkage`. |
| Markdown cells, tables, written analysis | They are plain text inside the notebook. |
| My GitHub repo | It is independent of either platform; I just keep pushing the exported `.ipynb` and PDF. |

### What needs redoing

| Account | Create a Kaggle account (Colab uses my Google login). |

_Platform details above are from my understanding of Kaggle's documentation and should be confirmed in the Kaggle notebook settings before an actual migration._

### What I personally ran into moving from local Jupyter to Colab
One of the things that really stood out for me as I transitioned from using  jupyter to google colab was getting accustomed to a new user interface and organization of the notebooks. While the python code ran and executed similarly on both platforms, it took me a little while to adjust to executing my work within a web browser versus directly on my laptop. What I learned from this experience is that although two systems may provide the exact same service (in this case, an interactive notebook environment), there will likely be at least some adjustments you'll need to make if you switch between them, which is what I'd expect from Kaggle another browser based interface.

### One risk
**Hard-coded file paths.** The code runs without errors on both platforms, but if I forget to change the CSV path, the notebook fails at the first `read_csv`, or worse, silently reads an old copy of the file. To reduce it, I would define the data path once in a single variable at the top of each notebook, and run Restart and Run All on the new platform before submitting so the saved outputs match the code.
